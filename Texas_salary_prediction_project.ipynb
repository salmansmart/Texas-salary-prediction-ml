```python
data.drop(['AGENCY NAME','LAST NAME','FIRST NAME','MI','CLASS CODE',
       'ETHNICITY','GENDER',
       'MONTHLY','STATE NUMBER',
       'duplicated', 'multiple_full_time_jobs', 'combined_multiple_jobs',
       'summed_annual_salary', 'hide_from_search'],axis=1,inplace=True)
```

```python
data.head(4)
```

```python
data.rename({"CLASS TITLE":"CLASS_TITLE"},axis=1,inplace=True)
```

```python
data.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['CLASS_TITLE']=le.fit_transform(data['CLASS_TITLE'])
```

```python
data["STATUS"]=le.fit_transform(data["STATUS"])
```

```python
data.head()
```

```python
data["EMPLOY DATE"]=pd.to_datetime(data["EMPLOY DATE"])
```

```python
data["EMPLOY DATE"]=pd.to_numeric(data["EMPLOY DATE"])
```

```python
plt.figure(figsize=(20,25),facecolor='white')
plot_number=1
for column in data:
    if plot_number<=7:
        ax=plt.subplot(5,3,plot_number)
        sns.boxplot(data[column])
        plt.xlabel(column,fontsize=20)
    plot_number+=1
plt.tight_layout
```

```python
from scipy.stats.mstats import winsorize
```

```python
trim_percentage=0.05
data["ANNUAL"]=winsorize(data['ANNUAL'],limits=trim_percentage)
```

```python
x=data[["AGENCY","CLASS_TITLE","STATUS","EMPLOY DATE","HRLY RATE","HRS PER WK"]]
```

```python
y=data["ANNUAL"]
```

```python
sns.heatmap(data.corr(),annot=True)
plt.show()
```

```python

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.fit_transform(x_test)
```

```python
x_train
```

```python
y_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model1=LinearRegression()
model1.fit(x_train,y_train)
```

```python
y1_pred=model1.predict(x_test)
y1_pred
```

```python
from sklearn.metrics import r2_score
report1=r2_score(y_test,y1_pred)
print(f"R2 Score: {report1}")
```

```python
from sklearn.tree import DecisionTreeRegressor
model2=DecisionTreeRegressor()
model2.fit(x_train,y_train)
```

```python
y2_pred=model2.predict(x_test)
```

```python
report2=r2_score(y_test,y2_pred)
print(report2)
```

```python
param={
       "criterion":["squared_error"],
       "splitter":("best","random"),
       "max_depth":(list(range(1,20))),
       "min_samples_split":[2,3,4],
       "min_samples_leaf":list(range(1,20))
      }
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(DecisionTreeRegressor(),param,verbose=1,n_jobs=-1,scoring="r2",cv=5)
grid.fit(x_train,y_train)
```

```python
print(grid.best_params_)
```

```python
model=DecisionTreeRegressor(criterion='squared_error', max_depth=19,min_samples_leaf=3,min_samples_split=4,splitter='best')
model.fit(x_train,y_train)
```

```python
y2_predict=model.predict(x_test)
y2_predict
```

```python
from sklearn.metrics import mean_squared_error
report_2=mean_squared_error(y_test,y2_predict)
print(report_2)
```

```python
report_dt=r2_score(y2_predict,y_test)
print(report_dt)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model3=RandomForestRegressor()
model3.fit(x_train,y_train)
y3_pred=model3.predict(x_test)
report4=r2_score(y_test,y3_pred)
print(report4)
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
grid_param={"n_estimators":[100,200,300,400,500],
      "max_features":["auto","sqrt"],
      "max_depth":(list(range(1,15))),
      "min_samples_split":[2,4,5],
      "min_samples_leaf":[1,2,4]}
```

```python
grid2=RandomizedSearchCV(model3,param_distributions=grid_param,n_iter=100,n_jobs=-1,verbose=1,scoring="r2",cv=3,refit=True)
```

```python
grid2.fit(x_train,y_train)
```

```python
pred_rf=grid2.predict(x_test)
pred_rf
```

```python
report3=r2_score(y_test,pred_rf)
print(report3)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
model4=GradientBoostingRegressor()
model4.fit(x_train,y_train)
y4_pred=model4.predict(x_test)
report4=r2_score(y_test,y4_pred)
print(report4)
```

```python
from xgboost import XGBRegressor
model5=XGBRegressor()
model5.fit(x_train,y_train)
y5_pred=model5.predict(x_test)
report5=r2_score(y_test,y5_pred)
print(report5)
```

```python
param5={"n_estimators":[100,200,300],
       "learning_rate":[0.001,0.005,0.1],
       "max_depth":[3,4,5],
       "min_child_weight":[5,6,7,8],
       "gamma":[0,0.1,0.2]}
```

```python
grid2=RandomizedSearchCV(model5,param_distributions=param5,n_iter=100,n_jobs=-1,verbose=1,scoring="r2",cv=3)
```

```python
grid2.fit(x_train,y_train)
```

```python
print(grid2.best_params_)
```

```python
model_xgb=XGBRegressor(n_estimators=300,min_child_weight=5,max_depth=5,learning_rate=0.1,gamma=0.1)
model_xgb.fit(x_train,y_train)
y5_predict=model_xgb.predict(x_test)
report5=r2_score(y_test,y5_predict)
print(report5)
```

```python
from sklearn.neural_network import MLPRegressor
model6=MLPRegressor(hidden_layer_sizes=(100,20),
                    learning_rate_init=(0.01),
                    max_iter=100,random_state=42)
model6.fit(x_train,y_train)
```
